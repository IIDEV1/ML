# Практическая работа №4. Регрессионная модель

**Уровень B.** Датасет IBM Telco Customer Churn.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
df=pd.read_csv("data/Telco-Customer-Churn.csv")
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors="coerce")
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
w=df[["tenure","MonthlyCharges","TotalCharges"]].dropna(); X=w[["tenure","MonthlyCharges"]]; y=w.TotalCharges; a,b,c,d=train_test_split(X,y,test_size=.2,random_state=42); m=make_pipeline(StandardScaler(),Ridge()).fit(a,c); p=m.predict(b); print(mean_absolute_error(d,p),mean_squared_error(d,p)**.5,r2_score(d,p)); plt.scatter(d,p); plt.xlabel("fact"); plt.ylabel("prediction"); plt.show()
```

### Вывод

Результаты представлены таблицами и графиками выше. Обработка данных выполняется воспроизводимо с фиксированным `random_state=42`.
