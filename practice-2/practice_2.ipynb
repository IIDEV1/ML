```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
df=pd.read_csv("data/Telco-Customer-Churn.csv")
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors="coerce")
num=["tenure","MonthlyCharges","TotalCharges"]; print(df[num].describe()); fig,ax=plt.subplots(1,3,figsize=(15,4)); [sns.boxplot(data=df,x=c,ax=a) for c,a in zip(num,ax)]; plt.show(); sns.heatmap(df[num].corr(),annot=True,cmap="Blues"); plt.show()
```
