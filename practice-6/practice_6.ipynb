# Практическая работа №6. Классификация текстов

**Уровень B.** Датасет IBM Telco Customer Churn.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
df=pd.read_csv("data/Telco-Customer-Churn.csv")
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors="coerce")
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
text=df[["Contract","InternetService","PaymentMethod","TechSupport","OnlineSecurity"]].fillna("unknown").astype(str).agg(" ".join,axis=1); y=df.Churn.map({"Yes":1,"No":0}); a,b,c,d=train_test_split(text,y,test_size=.2,stratify=y,random_state=42); m=make_pipeline(TfidfVectorizer(ngram_range=(1,2)),LogisticRegression(max_iter=1000)).fit(a,c); print(classification_report(d,m.predict(b)))
```

### Вывод

Результаты представлены таблицами и графиками выше. Обработка данных выполняется воспроизводимо с фиксированным `random_state=42`.
